````python
import pdfplumber
import pandas as pd
from IPython.display import display, Markdown
import os
````

````python
# sample pdf for testing
pdf_path = "sample.pdf"

if os.path.exists(pdf_path):
    with pdfplumber.open(pdf_path) as pdf:
        display(Markdown(f"### 📝 Full Page Text Extraction ({len(pdf.pages)} pages detected)"))
        
        for i, page in enumerate(pdf.pages):
            text = page.extract_text()
            display(Markdown(f"**--- Page {i+1} ---**"))
            print(text)
else:
    print(f" ERROR: File '{pdf_path}' not found.")
````

### 📊 Table Extraction

````python
if os.path.exists(pdf_path):
    with pdfplumber.open(pdf_path) as pdf:
        for i, page in enumerate(pdf.pages):
            # Extracting tables
            tables = page.extract_tables()
            
            if tables:
                display(Markdown(f"### Found {len(tables)} tables on Page {i+1}"))
                for j, table in enumerate(tables):
                    df = pd.DataFrame(table[1:], columns=table[0])
                    display(Markdown(f"**Table {j+1}:**"))
                    display(df)
            else:
                display(Markdown(f"*Page {i+1} has no detected tables.*"))
````
